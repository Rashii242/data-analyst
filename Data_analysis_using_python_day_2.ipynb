```python
# a dataframe from a dictionary of columns
students= pd.DataFrame({
    "name":["aarav","diya","kabir","meera","rohan"],
     "branch":["CSE","CSE","ECE","CSE","ECE"],
    "marks":[80,60,np.nan,77,90],
    "hourse_studied":[6,4,3,5,2],
})
students
```

```python
print("shape :",students.shape) # rows and columns
print("columns :",students.columns)
print("dtypes :\n",students.dtypes)
print()
students.info()
```

```python
print()
display(students[(students["branch"]=="CSE") | (students["marks"]>80)])
```

```python
# work on a copy keep the original
df=students.copy()
```

```python
df["marks_pct"]= df["marks"]/100 #arithmetic on a whole columns
print(df)
```

```python
df["passed"]=df["marks"]>=70 # boolean columns
print(df)
```

```python
df["name_upper"] =df["name"].str.upper() # .str gives vectorised text me
print(df)
```

```python
df["name_lower"] =df["name"].str.lower()
print(df)
```

```python
df.sort_values("marks", ascending=False)
```

```python
df.sort_values("marks")
```

```python
df["grade"]= pd.cut(df["marks"],
                    bins=[0,69,79,89,100],
                    labels=["D","C","B","A"])
print(df)
```

```python
df.isna().sum()  #missing per columns
```

```python
df.isna().any(axis=1).sum()  # rows with any missing
```

```python
df.isna().any(axis=0).sum()
```

```python
filled=df["marks"].fillna(df["marks"].median())     # replace with median
df["marks"].median()
print(filled.tolist())
```

```python
filled=df["marks"].fillna(df["marks"].mean())
df["marks"].mean()
print(filled.tolist())
```
